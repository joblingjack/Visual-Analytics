##Describe and Group By

These are important to note, especially when conducting EDA.

The ***describe()*** function in pandas provides descriptive statistics of a DataFrame or a Series. It gives a summary of key statistics for each numerical column in your data. It calculates summary statistics like count, mean, standard deviation, minimum, 25th percentile, median (50th percentile), 75th percentile, and maximum. It is applicable to numerical columns by default (but can include categorical data by specifying the include parameter).

Why use it: To quickly understand the distribution and spread of your data.

The ***groupby()*** function in pandas allows you to split your data into groups based on some criteria. It enables you to perform computations separately on each group and combine the results back into a single DataFrame. It Groups data based on one or more columns and applies aggregate functions such as sum, mean, count, max, etc., to each group.

Why use it: To analyze subgroups within your dataset, compute aggregated values, and gain insights into specific groups.

Run a describe() on the ***mtcars*** dataframe and group by the ***cyl*** variable.

```python
mtcars.info()
```

```python
#change data type to category and verify
#cyl: Number of cylinders: 4,6,8, vs:Engine: 0 = V-shaped, 1 = straight, am: Transmission: 0 = automatic, 1 = manual
mtcars['cyl'] = mtcars['cyl'].astype('category')
mtcars['vs'] = mtcars['vs'].astype('category')
mtcars['am'] = mtcars['am'].astype('category')

mtcars.info()
```

```python
mtcars.describe()
```

```python
mtcars.groupby('cyl').describe()
```

Create a summary statistics table using describe () of the data frame ***diamonds*** and group by ***cut***.

```python
diamonds.groupby('cut').describe()
```

If you want to only look at the ***carat*** variable from the code above.

```python
diamonds.groupby('cut')['carat'].describe()
```

In the context of the code you provided, the difference between parentheses () and square brackets [] is how they are used to access and manipulate data in pandas DataFrames:

Parentheses (): These are used to call functions or methods. For example, describe() is a method called on the grouped DataFrame to calculate descriptive statistics.
Square brackets []: These are used for indexing and selecting columns from a DataFrame. For example, ['carat'] selects the 'carat' column from the DataFrame or the grouped object. You can also use them to select multiple columns like ['carat', 'price'].
In the line diamonds.groupby('cut')['carat'].describe(), groupby('cut') is a method call using parentheses, ['carat'] is selecting a column using square brackets, and describe() is another method call using parentheses.

## Seaborn

We look at plotting with the ***seaborn*** library. ***Seaborn*** is a high-level data visualization library in Python built on top of Matplotlib. It is designed to make statistical data visualization more intuitive and attractive, providing an interface for drawing informative and aesthetically pleasing plots.

###Scatterplots
Create a scatterplot of the ***mtcars*** dataframe using x = ***hp*** and y = ***mpg***.

***sns.scatterplot():***  It is similar to plt.scatter() from Matplotlib but with additional features that make it easier to customize and make visually appealing plots.

***mtcars:*** This parameter specifies the dataset you are using for the plot. In this case, mtcars is the dataset, which refers to the classic dataset containing information about cars (horsepower, miles per gallon, weight, etc.).

***x="hp":*** This specifies the column name to use for the x-axis. In this case, "hp" stands for horsepower.

***y="mpg":*** This specifies the column name to use for the y-axis.

```python
sns.scatterplot(mtcars, x="hp", y="mpg")
```

Now lets color the points by the ***cyl*** variable.  

```python
sns.scatterplot(mtcars, x="hp", y="mpg", hue = "cyl")
```

If you don't like the colors, you can manually specify them.

```python
sns.scatterplot(mtcars, x="hp", y="mpg", hue = "cyl",
                palette=['green','orange','brown'])
```

```python
sns.scatterplot(mtcars, x="hp", y="mpg", hue = "cyl",
                    palette="coolwarm")
```
